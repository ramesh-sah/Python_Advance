# Simple pandas matplotlib data show

```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample data as dictionary
data = {
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "Sales": [120, 150, 170, 80, 200, 220, 180]
}

# Create DataFrame
df = pd.DataFrame(data)

# Line Plot
plt.figure(figsize=(8, 5))
plt.plot(df["Day"], df["Sales"], marker="o", linestyle="-", color="b", label="Sales")
plt.title("Weekly Sales Trend")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()

# Bar Plot
df.plot(x="Day", y="Sales", kind="bar", color="orange", legend=False, figsize=(8,5))
plt.title("Weekly Sales (Bar Chart)")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.show()

# Scatter Plot
df.plot(x="Day", y="Sales", kind="scatter", color="red", figsize=(8,5))
plt.title("Sales Scatter Plot")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.show()
```
